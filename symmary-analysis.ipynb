```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

## Static Stats Arbitrage

```python
df_static = pd.read_csv("statistical_arbitrage_backtest_summary.csv")
```

```python
df_static.info()
```

```python
plt.figure(figsize=(15, 4))
plt.bar(df_static['Pair'], df_static['Return (%)'], color='skyblue')
plt.axhline(0, color='red', linestyle='--', linewidth=1)
plt.title('Return by Pair', fontsize=16)
plt.xlabel('Pair', fontsize=14)
plt.ylabel('Return (%)', fontsize=14)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6), dpi=300)
plt.bar(df_static['Pair'], df_static['Half-Life (Days)'].sort_values(ascending=False), color='red')
plt.title('Half-Life by Pair', fontsize=16)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
